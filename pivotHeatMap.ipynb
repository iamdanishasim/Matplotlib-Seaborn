```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
flights = sns.load_dataset('flights')
flights
```

```python
flightscorr = flights[['year','passengers']]
```

```python
flightcorr = flightscorr.corr()
flightcorr
```

```python
sns.heatmap(flightcorr)
```

```python
flightspivot = flights.pivot_table(index='year',columns='month',values='passengers')
flightspivot
```

```python
sns.heatmap(flightspivot)
```
