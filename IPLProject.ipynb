```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
data = pd.read_csv('IPL.csv')
```

```python
data.head()
```

```python
data.shape
```

```python
data.isnull().sum()
```

```python
match_wins = data['match_winner'].value_counts()
match_wins
sns.barplot(y = match_wins.index , x = match_wins.values )
```

```python
toss_decision = data['toss_decision']
toss_decision
sns.countplot(toss_decision)
```

```python
won = data[data['toss_winner'] == data['match_winner']]['match_id'].count()
percent = won*100/data.shape[0]
print(percent)
```

```python
player_match = data['player_of_the_match'].value_counts().head(10)
player_match
sns.barplot(y = player_match.index , x = player_match.values)
#
```

```python
data['highest_wickets'] = data['best_bowling_figure'].apply(lambda x :x.split('--')[0])
data['highest_wickets'] = data['highest_wickets'].astype(int)
top_bowlers = data.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending=False)
top_bowlers.plot(kind='barh')
```

```python
venue_count = data['venue'].value_counts()
venue_count

sns.barplot(x=venue_count.index, y=venue_count.values)
```
